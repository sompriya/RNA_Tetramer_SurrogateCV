```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.interpolate

def plot_2d_fes(filename,max_fes,n_contour,xlabel='x',ylabel='y',cbar_label='Free Energy (kJ/mol)',cmap='coolwarm'):
    N = 200 #number of points for plotting/interpolation
    x, y, z = np.genfromtxt(filename, unpack=True)

    xi = np.linspace(x.min(), x.max(), N)
    yi = np.linspace(y.min(), y.max(), N)
    zi = scipy.interpolate.griddata((x, y), z, (xi[None,:], yi[:,None]))

    X, Y = np.meshgrid(xi, yi)

    cp = plt.contourf(X, Y, zi, levels=np.linspace(0,max_fes,n_contour-1), cmap=cmap)
    plt.clim(0.0,max_fes)
    cbar = plt.colorbar(cp)
    cbar.set_label(cbar_label,fontsize=16)
    plt.xlabel(xlabel,fontsize=16)
    plt.ylabel(ylabel,fontsize=16)
    


plot_2d_fes('mean_values.dat',40,20,xlabel=r'Deep-TICA_lasso1',ylabel=r'Deep-TICA_lasso2',cmap='fessa')

plt.xlim(-0.75,1.2)
plt.ylim(-1.2,0.75)
#plt.savefig('fes_our1.png',dpi=300,bbox_inches='tight')
plt.show()
```

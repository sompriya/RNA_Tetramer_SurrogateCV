```python
#DE Shaw
import numpy as np
import matplotlib.pyplot as plt
from mlcolvar.utils.plot import plot_metrics
from mlcolvar.utils.io import load_dataframe
from mlcolvar.utils.fes import compute_fes
import pandas as pd
df = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/DE_Shaw_analysis_mod_final/COLVAR_torsion_deshawCV',stride=1)


cv = df.filter(regex='lassoCV_node0|lassoCV_node1').values

fes_deS,grid_deS,bounds_deS,error_deS = compute_fes(cv,				                    
                                    blocks=1,
                                    bandwidth=0.01, scale_by='range',
                                    plot=False, eps=1e-10)
```

```python
pp = plt.contourf(fes_deS, cmap="fessa", vmax=20, levels=[2*i for i in range(11)],extent=(grid_deS[0][0,:].min(),grid_deS[0][0,:].max(),grid_deS[1][:,0].min(),grid_deS[1][:,0].max()))
cbar = plt.colorbar(pp)
plt.xlim(-0.9,1.2)
plt.ylim(-1.2,0.75)
plt.xlabel("Deep-TICA Lasso1")  # Set x-axis label
plt.ylabel("Deep-TICA Lasso2") 
cbar.set_label("FES")
plt.savefig('fes_ourcv_indeshaw_mod_aaaa_deshawCV.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mlcolvar.utils.plot import plot_metrics
from mlcolvar.utils.io import load_dataframe
from mlcolvar.utils.fes import compute_fes

#df = load_dataframe('COLVAR',start=50000,stop=1279880,stride=1)
df = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/OPES_lasso_10ns_mod_final/COLVAR',stride=1,start=100000,stop=1000000)

bias = df.filter(regex='opes.bias').values.sum(axis=1)

cv = df.filter(regex='lassoCV_node0|lassoCV_node1').values

# Calculate inverse temperature
temp=275.0
kb=0.008314
beta=1./(kb*temp)

# Compute log-weights for reweighting
logweights = beta*bias
w = np.exp(logweights)

fig,ax = plt.subplots(1,1,figsize=(6,5),dpi=100)



fes_10ns,grid_10ns,bounds_10ns,error_10ns = compute_fes(cv,
				                    weights=w,
                                    blocks=1,
                                    bandwidth=0.01, scale_by='range',
                                    plot=True, plot_max_fes=20, ax = ax, eps=1e-10)

ax.set_xlabel('Deep-TICA_lass0 1')
ax.set_ylabel('Deep-TICA_lasso 2')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

l1 = np.loadtxt('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/OPES_lasso_10ns_mod_final/deltaF.dat')
l2 = np.loadtxt('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/OPES_lasso_9.7ns_mod_final/deltaF.dat')
l3 = np.loadtxt('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/OPES_lasso_9.9ns_mod_final/deltaF.dat')

print(l1.shape)
print(l2.shape)
print(l3.shape)

# Multiplying the arrays by -1
#l1 = l1 * -1
#l2 = l2 * -1
#l3 = l3 * -1

Fmean = np.mean(np.vstack((l1,l2,l3)),axis=0)
print(Fmean.shape)

Fstd = np.std(np.vstack((l1,l2,l3)),axis=0)

xx = np.arange(len(Fmean))*0.01


plt.figure(figsize=(6,2))
plt.ylim(-20,10)


plt.fill_between(xx,Fmean-Fstd,Fmean+Fstd,color='xkcd:lightgreen',alpha=0.5)

#plt.plot(xx,l1,color='g',lw=1)
#plt.plot(xx,l3,color='g',lw=1)
#plt.plot(xx,l2,color='g',lw=1)
plt.plot(xx,Fmean,color='g',lw=1)

#reference = 4.1855*0.9 #kJ/mol
reference = -7.55
plt.plot(xx,np.ones(len(xx))*reference,'k--',label='Reference')
plt.plot(xx,np.ones(len(xx))*(reference+2.5),color='k',ls=':',label='Reference')
plt.plot(xx,np.ones(len(xx))*(reference-2.5),color='k',ls=':')
plt.xlabel('Simulation Time ($\mu$s)',fontsize=16)
plt.ylabel('$\Delta F$ (kJ/mol)',fontsize=16)

plt.xticks(fontsize=14)
#plt.xticks([0, 0.75, 1.5, 2.25], ['0.25', '1.0', '1.75', '2.50'], fontsize=14)
plt.xticks([0, 0.3, 0.6, 0.9], ['0.1', '0.4', '0.7', '1'], fontsize=14)
plt.yticks(fontsize=14)

plt.title('Surrogate Model CV',fontsize=16)
plt.savefig('AAAA_conv_deshawCV.png',dpi=300,bbox_inches='tight')
```

```python
#DE Shaw
import numpy as np
import matplotlib.pyplot as plt
from mlcolvar.utils.plot import plot_metrics
from mlcolvar.utils.io import load_dataframe
from mlcolvar.utils.fes import compute_fes
import pandas as pd
df = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/DE_Shaw_analysis_mod_final/COLVAR_NN',stride=1)


cv = df.filter(regex='DeepTICA.node-0|DeepTICA.node-1').values

fes_deS_NN,grid_deS_NN,bounds_deS_NN,error_deS_NN = compute_fes(cv,				                    
                                    blocks=1,
                                    bandwidth=0.01, scale_by='range',
                                    plot=False, eps=1e-10)

pp = plt.contourf(fes_deS_NN, cmap="fessa", vmax=20, levels=[2*i for i in range(11)],extent=(grid_deS_NN[0][0,:].min(),grid_deS_NN[0][0,:].max(),grid_deS_NN[1][:,0].min(),grid_deS_NN[1][:,0].max()))
cbar = plt.colorbar(pp)
plt.xlim(-0.9,1.2)
plt.ylim(-1.2,0.75)
cbar.set_label("FES")
plt.xlabel("Deep-TICA 1")  # Set x-axis label
plt.ylabel("Deep-TICA 2") 
plt.savefig('fes_ourcv_indeshaw_mod_aaaa_deshawCV_NN.png',dpi=300,bbox_inches='tight')
plt.show()
```

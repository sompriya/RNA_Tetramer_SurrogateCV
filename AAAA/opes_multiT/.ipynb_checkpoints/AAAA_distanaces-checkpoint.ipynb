```python
import numpy as np
import matplotlib.pyplot as plt
from mlcolvar.utils.plot import plot_metrics
from mlcolvar.utils.io import load_dataframe
from mlcolvar.utils.fes import compute_fes
import pandas as pd


# load colvar files containing weights of the OPES simulations as well as the driver with the input features
opes = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/opes_frm_multiT_lowT_withoutstatic_1us_10ns/COLVAR',
    start=100000, stop=1000000, stride=10)

feat = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/AAAA/DESRES-Trajectory_pnas2018-tetranucleotide-AAAA-no-water/pnas2018-tetranucleotide-AAAA-no-water/large_WBOX/opes_frm_multiT_lowT_withoutstatic_1us_10ns/COLVAR_dist',
    start=10000, stop=100000, stride=1)



bias = opes.filter(regex='.bias').values.sum(axis=1)

cv = feat.filter(regex='d1|d4').values

# Calculate inverse temperature
temp=275.0
kb=0.008314
beta=1./(kb*temp)

# Compute log-weights for reweighting
logweights = beta*bias
w = np.exp(logweights)

fig,ax = plt.subplots(1,1,figsize=(6,5),dpi=100)



fes_10ns_multiT_d1_d4,grid_10ns_multiT_d1_d4,bounds_10ns_multiT_d1_d4,error_10ns_multiT_d1_d4 = compute_fes(cv,
				                    weights=w,
                                    blocks=1,
                                    bandwidth=0.01, scale_by='range',
                                    plot=True, plot_max_fes=20, ax = ax, eps=1e-10)

#ax.set_xlabel('Deep-TICA_lass0 1')
#ax.set_ylabel('Deep-TICA_lasso 2')
plt.show()
```

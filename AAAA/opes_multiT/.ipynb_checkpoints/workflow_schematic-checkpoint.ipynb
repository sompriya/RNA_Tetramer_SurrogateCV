```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
l = np.loadtxt('DELTAFS')
temps = [275., 283.34, 294., 304., 314.32, 325]

```

```python
kb = 0.00833
temp_vs_time = []
for i in range(len(l)):
    exp_sum = 0.0
    numerator = 0.0
    for j in range(len(temps)):
        T = temps[j]
        exp_sum += np.exp(-l[i,j+2])
        numerator += T*np.exp(-l[i,j+2])
    temp_vs_time.append(numerator)
print(temp_vs_time)
```

```python
plt.plot(np.arange(len(l)),temp_vs_time)
```

```python
import numpy as np
import matplotlib.pyplot as plt
l = np.loadtxt('COLVAR_notReal')
```

```python
maxT = 340
minT = 270
xx = l[:,0]
yy = l[:,1]

new_y = minT + (maxT - minT)*yy
```

```python
plt.scatter(xx,new_y,s=15,color='orange',alpha=0.9)
plt.axhline(y=275,ls='--',color='k',label='275 K',lw=3)
plt.axhline(y=285,ls='--',color='k',label='285 K',lw=3)
plt.axhline(y=295,ls='--',color='k',label='295 K',lw=3)
plt.axhline(y=305,ls='--',color='k',label='305 K',lw=3)
plt.axhline(y=315,ls='--',color='k',label='315 K',lw=3)
plt.axhline(y=325,ls='--',color='k',label='325 K',lw=3)
#plt.legend()
#plt.legend(prop={'weight': 'bold'})
plt.text(3000,278,'275 K',fontsize=20)
plt.text(3000,288,'285 K',fontsize=20)
plt.text(0,298,'295 K',fontsize=20)
plt.text(0,308,'305 K',fontsize=20)
plt.text(0,318,'315 K',fontsize=20)
plt.text(0,328,'325 K',fontsize=20)
plt.tick_params(
    axis='both',          # Changes apply to both x and y axes
    which='both',      # Both major and minor ticks are affected
    bottom=False,      # Ticks along the bottom edge are off
    top=False,         # Ticks along the top edge are off
    left=False,        # Ticks along the left edge are off
    right=False,       # Ticks along the right edge are off
    labelbottom=False, # Labels along the bottom edge are off
    labelleft=False    # Labels along the left edge are off
)
plt.savefig('Workflow_schematic.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
#DE Shaw
import numpy as np
import matplotlib.pyplot as plt
from mlcolvar.utils.plot import plot_metrics
from mlcolvar.utils.io import load_dataframe
from mlcolvar.utils.fes import compute_fes
import pandas as pd
df = load_dataframe('/home/sompriya/Desktop/RNA_tetranucleotides/UUUU/DESRES-Trajectory_pnas2018-tetranucleotide-UUUU-no-water/pnas2018-tetranucleotide-UUUU-no-water/DeShaw_analysis/COLVAR_NN',stride=1)


cv = df.filter(regex='DeepTICA.node-0|DeepTICA.node-1').values

fes_deS_NN,grid_deS_NN,bounds_deS_NN,error_deS_NN = compute_fes(cv,				                    
                                    blocks=1,
                                    bandwidth=0.01, scale_by='range',
                                    plot=False, eps=1e-10)

pp = plt.contourf(fes_deS_NN, cmap="magma", edgecolor='blue', vmax=20, levels=[2*i for i in range(11)],extent=(grid_deS_NN[0][0,:].min(),grid_deS_NN[0][0,:].max(),grid_deS_NN[1][:,0].min(),grid_deS_NN[1][:,0].max()))
cbar = plt.colorbar(pp)
plt.xlim(-0.9,1.2)
plt.ylim(-1,0.9)
cbar.set_label("FES")
plt.xlabel("Deep-TICA 1")  # Set x-axis label
plt.ylabel("Deep-TICA 2")  
#plt.savefig('/home/sompriya/Desktop/fes_ourcv_indeshaw_mod_uuuu_deshawCV_NN1.png',dpi=300,bbox_inches='tight')
plt.show()
```
